# Exercise 1: Prompt Chaining for a Customer Support AI

This exercise uses multiple prompts to handle a customer complaint. Each step uses the output from the previous step.

## Step 1: Classify the Customer Complaint

Prompt used:

You are a customer support triage assistant.

Classify the following customer complaint into one category: billing, technical issue, account access, cancellation, or other.

Complaint:
“I was charged twice for my monthly subscription, and I want one charge refunded.”

Return your answer in this format:

Category:
Reason:

Keep the answer short and clear.

Output:

Category: Billing

Reason: The customer reports being charged twice and wants one charge refunded.

Tool used: Google Colab and ChatGPT. The AI responses were manually copied into Python because no API key was used.

```python
complaint = "I was charged twice for my monthly subscription, and I want one charge refunded."

step1_prompt = """
You are a customer support triage assistant.

Classify this customer complaint as billing, technical issue,
account access, cancellation, or other.

Complaint:
I was charged twice for my monthly subscription, and I want one charge refunded.

Return your answer in this format:
Category:
Reason:

Keep the answer short and clear.
"""

step1_response = """
Category: Billing
Reason: The customer reports being charged twice and wants one charge refunded.
"""

print("Customer Complaint:")
print(complaint)

print("\nPrompt Used:")
print(step1_prompt)

print("AI Response:")
print(step1_response)
```

## Step 2: Identify Missing Information

This step uses the customer complaint and the billing classification from Step 1 to determine what information is needed to investigate the issue.

Prompt used:

You are a customer support assistant.

Customer complaint:
“I was charged twice for my monthly subscription, and I want one charge refunded.”

Step 1 classification:
Category: Billing
Reason: The customer reports being charged twice and wants one charge refunded.

Based on the complaint and classification, list 3–5 pieces of information needed to investigate the issue.

Do not ask for a password or full credit card number.

Return only a numbered list.

Output:

1. The dates of both charges.
2. The amounts of both charges.
3. The email address connected to the subscription.
4. The name of the subscription plan.
5. A screenshot showing both charges.

```python
step2_prompt = f"""
You are a customer support assistant.

Customer complaint:
{complaint}

Step 1 classification:
{step1_response}

Based on the complaint and classification, list 3–5 pieces of information
needed to investigate the issue.

Do not ask for a password or full credit card number.
If requesting a screenshot, tell the customer to hide sensitive information.

Return only a numbered list.
"""

step2_response = """
1. The dates of both charges.
2. The amounts of both charges.
3. The email address connected to the subscription.
4. The name of the subscription plan.
5. A screenshot of both charges with sensitive information hidden.
"""

print("Step 2 Prompt:")
print(step2_prompt)

print("Step 2 - Missing Information:")
print(step2_response)
```

## Step 3: Propose a Solution

This step uses the complaint, classification, and missing information from the previous steps to recommend an appropriate solution.

Prompt used:

You are a billing support specialist.

Customer complaint:
“I was charged twice for my monthly subscription, and I want one charge refunded.”

Step 1 classification:
Category: Billing
Reason: The customer reports being charged twice and wants one charge refunded.

Step 2 missing information:
1. The dates of both charges.
2. The amounts of both charges.
3. The email address connected to the subscription.
4. The name of the subscription plan.
5. A screenshot showing both charges.

Based on this information, propose a solution.

Do not promise a refund before confirming that the charge is duplicated. Do not ask for a password or full credit card number.

Return your answer in this format:

Recommended action:
Information to verify:
Customer communication:

Keep the answer clear and professional.

Output:

Recommended action:
Verify both transactions and refund one charge if the duplicate charge is confirmed.

Information to verify:
Check the dates, amounts, subscription email, plan name, and proof of both charges.

Customer communication:
Explain that the charges will be reviewed and that one refund can be processed if the duplicate charge is confirmed.

```python
step3_prompt = f"""
You are a billing support specialist.

Customer complaint:
{complaint}

Classification:
{step1_response}

Missing information:
{step2_response}

Based on this information, propose a solution.

Do not promise a refund before confirming the duplicate charge.
Do not ask for a password or full credit card number.

Return your answer in this format:

Recommended action:
Information to verify:
Customer communication:

Keep the answer clear and professional.
"""

step3_response = """
Recommended action:
Verify both transactions and refund one charge if the duplicate is confirmed.

Information to verify:
Check the dates, amounts, subscription email, plan name, and proof of both charges.

Customer communication:
Explain that the charges will be reviewed and that one refund can be processed if the duplicate charge is confirmed.
"""

print("Step 3 Prompt:")
print(step3_prompt)

print("Step 3 - Proposed Solution:")
print(step3_response)
```

## Step 4: Escalation Decision

This step uses the previous classification, missing information, and proposed solution to decide whether the case should be escalated.

Prompt used:

You are a customer support escalation assistant.

Customer complaint:
“I was charged twice for my monthly subscription, and I want one charge refunded.”

Classification:
Category: Billing
Reason: The customer reports being charged twice and wants one charge refunded.

Missing information:
1. The dates of both charges.
2. The amounts of both charges.
3. The email address connected to the subscription.
4. The name of the subscription plan.
5. A screenshot showing both charges.

Proposed solution:
Verify both transactions and refund one charge if the duplicate charge is confirmed.

Decide whether this case should be handled by a standard support agent or escalated to a billing specialist.

Escalate only if the duplicate charge cannot be verified, the refund fails, or the case requires special approval.

Output:

Escalation decision:
Standard support process.

Reason:
A standard billing agent can verify the duplicate charge and process one refund. Escalation is only needed if the charge cannot be verified, the refund fails, or special approval is required.

Next step:
Verify both transactions and process one refund if the duplicate charge is confirmed.

```python
step4_prompt = f"""
You are a customer support escalation assistant.

Customer complaint:
{complaint}

Classification:
{step1_response}

Missing information:
{step2_response}

Proposed solution:
{step3_response}

Decide whether this case should be handled by a standard support agent
or escalated to a billing specialist.

Escalate only if the duplicate charge cannot be verified, the refund fails,
or special approval is required.

Return your answer in this format:

Escalation decision:
Reason:
Next step:
"""

step4_response = """
Escalation decision:
Standard support process.

Reason:
A standard billing agent can verify the duplicate charge and process one refund. Escalation is only needed if the charge cannot be verified, the refund fails, or special approval is required.

Next step:
Verify both transactions and process one refund if the duplicate charge is confirmed.
"""

print("Step 4 Prompt:")
print(step4_prompt)

print("Step 4 - Escalation Decision:")
print(step4_response)
```

## Step 5: Critique and Improved Response

This step tests and improves the proposed solution from Step 3 and the escalation decision from Step 4.

Prompt used:

You are a customer support quality reviewer.

Customer complaint:
“I was charged twice for my monthly subscription, and I want one charge refunded.”

Classification:
Billing issue.

Missing information:
The dates and amounts of both charges, the subscription email, the plan name, and proof of both charges.

Proposed solution:
Verify both transactions and refund one charge if the duplicate charge is confirmed.

Escalation decision:
Use the standard support process. Escalate only if the charge cannot be verified, the refund fails, or special approval is required.

Critique the proposed solution and escalation decision. Then write an improved customer-facing response.

The improved response must:
- Be empathetic and professional
- Clearly explain what information is needed
- Not request a password or full credit card number
- Not promise a refund before verification
- Be less than 120 words

Output:

Critique:
The proposed solution is clear and appropriate because it explains that the duplicate charge must be verified before a refund is issued. The escalation rule is also reasonable because a standard billing agent can handle the case unless verification or refund processing fails.

Improved customer-facing response:
I’m sorry you were charged twice for your subscription. We can review the two transactions and process a refund if one charge is confirmed to be a duplicate. Please provide the dates and amounts of both charges, the email address connected to your subscription, and a screenshot showing the charges. Please do not send your password or full credit card number. If we cannot verify the duplicate charge or the refund cannot be processed, we will escalate the issue to a billing specialist.

```python
step5_prompt = f"""
You are a customer support quality reviewer.

Customer complaint:
{complaint}

Proposed solution:
{step3_response}

Escalation decision:
{step4_response}

Critique the solution and escalation decision. Then write an improved
customer-facing response.

The improved response must:
- Be empathetic and professional
- Clearly explain what information is needed
- Not request a password or full credit card number
- Not promise a refund before verification
- Be fewer than 120 words

Return your answer in this format:

Critique:
Improved customer-facing response:
"""

step5_response = """
Critique:
The proposed solution is clear because it requires verification before issuing a refund. The escalation decision is also appropriate because a billing agent can handle the issue unless verification or refund processing fails.

Improved customer-facing response:
I’m sorry you were charged twice for your subscription. Please provide the dates and amounts of both charges, your subscription email, and a screenshot with sensitive information hidden. We will review the transactions and process a refund if one charge is confirmed to be a duplicate. Please do not send your password or full credit card number. If the issue cannot be verified or the refund fails, we will escalate it to a billing specialist.
"""

print("Step 5 Prompt:")
print(step5_prompt)

print("Step 5 - Critique and Improved Response:")
print(step5_response)
```
