## ReACT Prompt Used

I need Python code that moves all zeros in a list to the end while keeping the other numbers in the same order.

Follow these steps:

1. Plan: Briefly explain the approach.
2. Generate: Write Python code using only basic Python.
3. Run: Test the code with [0, 1, 0, 3, 12].
4. Observe: Check whether the output is [1, 3, 12, 0, 0].
5. Fix: If the output is incorrect, revise the code and test it again.

Also test these lists:

[0, 0, 1]
[1, 2, 3]
[0, 5, 0, 7]

The code should preserve the original order of all nonzero numbers and should not use external libraries.

```python
def move_zeros(numbers):
    non_zero_numbers = [number for number in numbers if number != 0]
    zero_count = numbers.count(0)
    return non_zero_numbers + [0] * zero_count


test_lists = [
    [0, 1, 0, 3, 12],
    [0, 0, 1],
    [1, 2, 3],
    [0, 5, 0, 7],
    []
]

for numbers in test_lists:
    result = move_zeros(numbers)

    print("Input:", numbers)
    print("Output:", result)
    print()
```

## ReACT Process Results

Plan: I planned to separate the nonzero numbers from the zeros and then place the zeros at the end.

Generate: The Python code used a list comprehension and the count() function. It used only basic Python and no external libraries.

Run: I ran the code using the example list and additional test lists.

Observe: The outputs were correct. The nonzero numbers stayed in their original order, and all zeros moved to the end.

Fix: No correction was needed because the code worked correctly for every test case.

The code also handled an empty list correctly.
