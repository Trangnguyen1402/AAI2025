# Exercise 3: Self-Reflection Prompt for Improving Output

Tool Used:
Google Colab and ChatGPT.

Goal:
Ask the AI to review an original summary and improve it.

```python
original_summary = """
AI is used in customer service to answer common questions and help support agents.
It can save time and make customer service more efficient, but it can also make mistakes.
Human employees should review important or sensitive customer issues.
"""

print("Original Summary:")
print(original_summary)
```

```python
self_critique_prompt = f"""
Review the summary below.

Check it for:
- Accuracy
- Clarity
- Intended audience
- Focus
- Important missing information
- Length and organization

Explain what could be improved, then suggest specific changes.
Do not rewrite the summary yet.

Original summary:
{original_summary}
"""

self_critique_response = """
The summary is accurate and easy to understand, but it is somewhat general.
It could be improved by explaining how AI helps with common questions and reduces
employee workload. It should also mention that AI can make mistakes and that
human review is important for sensitive situations.
"""

print("Self-Critique Prompt:")
print(self_critique_prompt)

print("Self-Critique Response:")
print(self_critique_response)
```

```python
improvement_prompt = f"""
Rewrite the original summary using the critique.

Requirements:
- Keep the summary under 80 words.
- Make it clear and accurate.
- Explain how AI helps customer service.
- Mention that AI can make mistakes.
- Explain why human review is important.
- Use a professional but simple tone.

Original summary:
{original_summary}

Critique:
{self_critique_response}
"""

improved_summary = """
AI helps customer service by answering common questions and reducing employee
workload. However, it can make inaccurate or incomplete responses, especially
when handling sensitive issues. Human employees should review important cases
to ensure customers receive accurate, safe, and appropriate support.
"""

print("Improvement Prompt:")
print(improvement_prompt)

print("Improved Summary:")
print(improved_summary)
```

## Improvement Note

The revised summary is longer, clearer, and explains the importance of human review more fully. It still focuses on the main ideas and follows the under-80-word requirement.
